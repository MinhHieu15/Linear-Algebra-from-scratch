## Part D — Library: `is_zero_vector` and `vectors_close`

**Why:** later functions must check "is this vector zero?" before dividing by its length (Day 8), and almost every test compares two vectors. Both need a tolerance, as you saw in C3.

Both functions are already in `linalg/vectors.py` with a docstring and a `raise NotImplementedError` line. Replace that line with your code.

**Task D1 — `is_zero_vector(v, tol=1e-10)`:** return `True` if **every** entry of `v` has absolute value below `tol`.

Steps:
1. `v = as_vector(v)` (reuse your Day 2 function)
2. build the mask with `np.abs(...)` (Day 2, Task 3.3: don't forget `abs`!)
3. `np.all(mask)` is `True` only if every entry is `True`

Check it:

````python
from linalg.vectors import is_zero_vector

print(is_zero_vector([0, 0, 0]))         # expected: True
print(is_zero_vector([1e-12, -3e-11]))   # expected: True
print(is_zero_vector([0.0, 0.001]))      # expected: False
print(is_zero_vector([-5.0, 0.0]))       # expected: False  <- fails if you forgot abs
````

**Task D2 — `vectors_close(u, v, tol=1e-10)`:** return `True` if `u` and `v` have the same length and every pair of entries differs by less than `tol`.

Steps:
1. convert both with `as_vector`
2. if their shapes differ, `raise ValueError` with both shapes in the message
3. the difference `u - v` should be a zero vector, so you can reuse `is_zero_vector`!

If you edited the file after importing, click **Restart** at the top first. Check it:

````python
from linalg.vectors import vectors_close

print(vectors_close([0.1 + 0.2, 1.0], [0.3, 1.0]))  # expected: True
print(vectors_close([1.0, 2.0], [1.0, 2.1]))        # expected: False

try:
    vectors_close([1, 2], [1, 2, 3])
except ValueError as e:
    print("ValueError:", e)                         # expected: your message
````
