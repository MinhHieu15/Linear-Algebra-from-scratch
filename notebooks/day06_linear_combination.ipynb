**B2 - Write `linear_combination(coeffs, vectors)`:**

**Why:** "scale some vectors, then add them up" is the most important move in linear algebra.

Fill in `linear_combination` in `linalg/vectors.py`. Steps:
1. if `len(coeffs) != len(vectors)`, `raise ValueError("Need one coefficient per vector")`
2. start with `result = np.zeros(len(vectors[0]))`
3. loop over the pairs `(c, v)` (`zip` helps) and set `result = add(result, scale(c, v))`
4. return `result`
